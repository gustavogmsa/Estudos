```python
##Exercicio 06

x = input('Digite algo:')

print(f'É um dígito? {x.isdigit()}')
print(f'É uma letra? {x.isalpha()}')
print(f'É uma caractere alfanumérico? {x.isalnum()}')
print(f'É tudo minusculo? {x.islower()}')
print(f'É tudo maiusculo? {x.isupper()}')
print(f'É apenas um espaço? {x.isspace()}')
```
